```python
def bot_wake():
    print("Bot is awake")

def bot_sleep():
    print("Bot is asleep")

def run_pending_jobs():
    # Convert current UTC time to New York time
    utc_now = datetime.now(pytz.utc)
    ny_now = utc_now.astimezone(pytz.timezone('America/New_York'))
    
    print(str(utc_now))
    # Only run pending jobs if it's between 8:30 AM and 5:00 PM New York time
    if ny_now.time() >= Time(8, 30) and ny_now.time() < Time(17, 0):
        print("Working my shift...")
        schedule.run_pending()

# Schedule the 'wake up' and 'sleep' tasks
schedule.every().day.at("13:30").do(bot_wake)  # 8:30 AM New York time in UTC
schedule.every().day.at("22:00").do(bot_sleep)  # 5:00 PM New York time in UTC

while True:
    run_pending_jobs()
    time.sleep(10)
```
